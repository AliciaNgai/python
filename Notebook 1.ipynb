# Excercise 7 - If/else/elif

```python
def weather():
    temp = float(input("what's the temperature in degrees celcius?"))
    if temp <= 0:
        print("it's freezing cold")
    elif temp <= 15:
        print("it's cold")
    elif temp <= 30:
        print("it's hot")
    elif temp <= 45:
        print("it's really hot")
    elif temp > 45:
        print("it's extremely hot")
```

```python
weather()
```
